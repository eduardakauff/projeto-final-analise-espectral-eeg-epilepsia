# 2. Importar as bibliotecas

```python
import pyedflib
import os
import glob
import pandas as pd
```

```python
arquivos = ["chb01_01.edf"]
```

# 4. Extrair os metadados de aquisição

```python
metadados = []

for arquivo in arquivos:

    edf = pyedflib.EdfReader(arquivo)

    nome_arquivo = os.path.basename(arquivo)

    # Informações gerais
    fs = edf.getSampleFrequency(0)
    duracao = edf.getFileDuration()
    numero_canais = edf.signals_in_file

    print("\n===================================")
    print("Arquivo:", nome_arquivo)
    print("Frequência de amostragem:", fs, "Hz")
    print("Duração:", duracao, "segundos")
    print("Número de canais:", numero_canais)
    print("===================================")

    # Informações individuais de cada canal
    for i in range(numero_canais):

        canal = edf.getLabel(i)

        unidade = edf.getPhysicalDimension(i)

        pmin = edf.getPhysicalMinimum(i)
        pmax = edf.getPhysicalMaximum(i)

        dmin = edf.getDigitalMinimum(i)
        dmax = edf.getDigitalMaximum(i)

        # Fator de conversão
        ganho = (pmax - pmin) / (dmax - dmin)

        # Resolução digital do sinal
        bits = (dmax - dmin + 1).bit_length()

        print(f"\nCanal: {canal}")
        print("Unidade:", unidade)
        print("Limite físico mínimo:", pmin)
        print("Limite físico máximo:", pmax)
        print("Limite digital mínimo:", dmin)
        print("Limite digital máximo:", dmax)
        print("Fator de conversão:", ganho)
        print("Resolução digital estimada:", bits, "bits")

        metadados.append({
            "Arquivo": nome_arquivo,
            "Canal": canal,
            "Frequência (Hz)": fs,
            "Duração (s)": duracao,
            "Unidade": unidade,
            "Limite físico mínimo": pmin,
            "Limite físico máximo": pmax,
            "Limite digital mínimo": dmin,
            "Limite digital máximo": dmax,
            "Fator de conversão": ganho,
            "Resolução estimada (bits)": bits
        })

    edf.close()
```

# 5. Organizar os resultados em uma tabela

```python
tabela = pd.DataFrame(metadados)

display(tabela)
```

# 6. Resultados

```python
# Resumo dos metadados de aquisição

arquivo = "chb01_01.edf"

edf = pyedflib.EdfReader(arquivo)

# Informações gerais
fs = edf.getSampleFrequency(0)
duracao = edf.getFileDuration()
numero_canais = edf.signals_in_file

# Informações do primeiro canal
pmin = edf.getPhysicalMinimum(0)
pmax = edf.getPhysicalMaximum(0)
dmin = edf.getDigitalMinimum(0)
dmax = edf.getDigitalMaximum(0)
unidade = edf.getPhysicalDimension(0)

# Fator de conversão
fator_conversao = (pmax - pmin) / (dmax - dmin)

# Resolução digital estimada
bits = (dmax - dmin + 1).bit_length()

# Exibição
print("METADADOS DE AQUISIÇÃO - CHB01")
print("-----------------------------------")
print(f"Arquivo: {arquivo}")
print(f"Número de canais: {numero_canais}")
print(f"Frequência de amostragem: {fs} Hz")
print(f"Duração: {duracao / 3600:.2f} horas")
print(f"Unidade: {unidade}")
print(f"Limites físicos: {pmin} a {pmax} {unidade}")
print(f"Limites digitais: {dmin} a {dmax}")
print(f"Resolução digital estimada: {bits} bits")
print(f"Fator de conversão: {fator_conversao:.5f} {unidade}/unidade digital")

edf.close()
```
